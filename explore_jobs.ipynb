```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("jobs.csv")
df["region"] = df["location"].str.split(",").str[-1].str.strip()
df.shape
```

```python
df.head(10)
```

```python
df["region"].value_counts()
```

```python
counts = df["region"].value_counts()
counts.plot(kind="pie", autopct="%1.1f%%", figsize=(5, 5), title="Jobs per region")
plt.ylabel("")
plt.show()
```

```python
is_python = df["title"].str.contains("python", case=False)
is_python.value_counts().rename({True: "Python", False: "Other"}).plot(
    kind="pie", autopct="%1.1f%%", figsize=(5, 5), title="Share of Python jobs")
plt.ylabel("")
plt.show()
```
